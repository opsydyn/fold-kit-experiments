# foldkit-viz in Jupyter
A plot spec drawn by foldkit-viz, checked against its JSON Schema. It follows the notebook's light or dark theme.

```python
import math
from foldkit_viz import Plot, PlotSpecError

rows = [{"x": i, "y": math.sin(i / 5) * 10 + i / 3, "group": ["alpha", "beta", "gamma"][i % 3]} for i in range(40)]
plot = Plot({
    "width": 520,
    "height": 300,
    "title": "Sine drift",
    "data": rows,
    "marks": [
        {"mark": "line", "x": "x", "y": "y", "stroke": "group"},
        {"mark": "dot", "x": "x", "y": "y", "fill": "group", "title": "group"},
    ],
})
plot
```

Assigning a new spec redraws the chart above.

```python
plot.spec = {**plot.spec, "title": "Sine drift, redrawn", "marks": plot.spec["marks"][1:]}
```

The schema catches a misspelt key before anything is drawn.

```python
try:
    Plot({**plot.spec, "titel": "typo"})
except PlotSpecError as error:
    print(error)
```

A field missing from the data passes the schema. foldkit-viz reports it in place of the chart, and in `error` once the chart has been displayed.

```python
missing = Plot({**plot.spec, "marks": [{"mark": "dot", "x": "dose", "y": "y"}]})
missing
```
