# foldkit-viz in a TypeScript notebook
Run with Deno's Jupyter kernel (`deno jupyter --install`). A cell whose value is `displayPlot(spec)` shows the chart inline, in the notebook's light or dark theme.

This imports the repository's build. Once released, import `npm:@opsydyn/foldkit-viz/jupyter` and `npm:@opsydyn/foldkit-viz/plot` instead.

```typescript
import { displayPlot, plotMimeBundle } from '../dist/jupyter.mjs';
import { columnTable, compilePlot } from '../dist/plot.mjs';

type Sample = { x: number; y: number; group: string };
const rows: Sample[] = Array.from({ length: 40 }, (_, i) => ({
  x: i,
  y: Math.sin(i / 5) * 10 + i / 3,
  group: ['alpha', 'beta', 'gamma'][i % 3]!,
}));

displayPlot({
  width: 520,
  height: 300,
  title: 'Sine drift',
  data: rows,
  marks: [
    { mark: 'line', x: 'x', y: 'y', stroke: 'group' },
    { mark: 'dot', x: 'x', y: 'y', fill: 'group', title: 'group' },
  ],
});
```

Column data, as typed arrays, and a polar chart with a continuous colour legend.

```typescript
const directions = ['N', 'NE', 'E', 'SE', 'S', 'SW', 'W', 'NW'];
displayPlot({
  width: 320,
  height: 320,
  title: 'Wind rose',
  coordinates: 'polar',
  data: columnTable({ dir: directions, speed: Float64Array.from([4, 6, 7, 5, 3, 2, 5, 6]) }),
  marks: [{ mark: 'bar', x: 'dir', y: 'speed', fill: 'speed' }],
});
```

A spec error is raised in the cell that calls `displayPlot`, naming the spec path.

```typescript
try {
  displayPlot({ width: 300, height: 200, data: rows, marks: [{ mark: 'dot', x: 'dose', y: 'y' }] });
} catch (error) {
  console.log((error as Error).message);
}
```

`plotMimeBundle` builds the same bundle from compiled geometry, for `Deno.jupyter.display`.

```typescript
const geometry = compilePlot({
  width: 420,
  height: 220,
  title: 'Fruit',
  data: [
    { fruit: 'apples', count: 5 },
    { fruit: 'pears', count: 3 },
    { fruit: 'plums', count: 7 },
  ],
  marks: [{ mark: 'bar', x: 'fruit', y: 'count' }],
});
await Deno.jupyter.display(plotMimeBundle(geometry, { summary: true }), { raw: true });
```
